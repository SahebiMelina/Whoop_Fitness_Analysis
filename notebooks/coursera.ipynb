In data analysis, we often convert categorical variable into dummy variable, if the observation belongs to the specified category, the dummy variable indicating the category would equal to 1, otherwise it equals to 0.

Convert a categorical variable to a dummy variable
The variable "gender" only carries two values, Female or Male. We will create dummy variables to capture the categories.

We can use the “pd.get_dummies” function to convert a categorical variable to dummy variable. This function will also omit any missing value.

```python
dummy=pd.get_dummies(whoop_fitness_dataset, columns=['gender'])
dummy.columns
```

Notice that two variables are created, gender_Female and gender_Male. The original variable gender is deleted.

We can attach the "gender_Male" and "gender_Female" dummy variables back to our whoop_fitness_dataset dataset using the pd.concat function.
axis=1 specifies that we are adding a column to the dataset;
axis=0 indicates adding rows

```python


whoop_fitness_dataset = pd.concat(
    [whoop_fitness_dataset, dummy[['gender_Male', 'gender_Female']]],
    axis=1
)

whoop_fitness_dataset.head()
```

```python
whoop_fitness_dataset.rename(columns={'gender_Male':'Male','gender_Female':'Female'}, inplace=True)
whoop_fitness_dataset.head()
```
