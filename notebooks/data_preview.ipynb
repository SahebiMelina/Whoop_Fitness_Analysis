# Whoop Fitness Dataset Preview

This notebook loads and displays the Whoop fitness dataset CSV file so that we can have a better understanding of all the information the file gives us !

Firstly We'll need to import the required libraries.

```python
import pandas as pd
import numpy as np 
import os
```

Secondly We'll need to locate the csv file so that it can be found by the system and in order to make sure if it has been found it gives us a binary response .

```python

csv_path = '../data/whoop_fitness_dataset_100k.csv'
print(f"Looking for CSV at: {csv_path}")
print(f"File exists: {os.path.exists(csv_path)}")
```

Then we'll load the data in a pandas dataframe and put its first 5 rows into display to be able to see how our data is organized .

```python

df = pd.read_csv(csv_path)
print(f"Dataset loaded successfully!")
print(f"Shape: {df.shape} (rows, columns)")
print("\n=== First 5 rows of data ===\n")
display(df.head())
```
